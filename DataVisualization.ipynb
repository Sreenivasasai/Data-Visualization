```python
languages = ['C','C++','Java','Python','R']
students = [200,250,350,400,150]
colors = ['green','red','purple','blue','orange']
```

```python
plt.figure(figsize=(7,5),facecolor='lightgrey')
plt.bar(languages,students,color=colors)
plt.title('Number of Students in different Languages')
plt.xlabel('Languages')
plt.ylabel('Number of Students')
ax = plt.gca()
ax.set_facecolor('lightblue')
```

```python
plt.barh(languages,students,color=colors)
plt.title('Number of Students in different Languages')
plt.ylabel('Languages')
plt.xlabel('Number of Students')
plt.show()
```

```python
explode = (0, 0, 0.05, 0.1,0)
pieplot = plt.pie(students,labels= languages,shadow=True,autopct='%1.2f%%',explode=explode ,startangle=120)
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(6, 3), subplot_kw=dict(aspect="equal"))

recipe = ["375 g flour",
          "75 g sugar",
          "250 g butter",
          "300 g berries"]

data = [float(x.split()[0]) for x in recipe]
ingredients = [x.split()[-1] for x in recipe]


def func(pct, allvals):
    absolute = int(np.round(pct/100.*np.sum(allvals)))
    return f"{pct:.1f}%\n({absolute:d} g)"


wedges, texts, autotexts = ax.pie(data, autopct=lambda pct: func(pct, data),
                                  textprops=dict(color="w"))

ax.legend(wedges, ingredients,
          title="Ingredients",
          loc="center left",
          bbox_to_anchor=(1, 0, 0.5, 1))

plt.setp(autotexts, size=8, weight="bold")

ax.set_title("Matplotlib bakery: A pie")

plt.show()
```
